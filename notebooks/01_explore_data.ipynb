# 01 Explore the CFPB complaints data
Steps: load, count classes, read samples per class, choose the metric.

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("liri-uzh/cfpb-complaints-mini")
train = ds["train"].to_pandas()
test = ds["test"].to_pandas()
train.shape, test.shape
```

```python
# Step 1: class counts
print(train["label"].value_counts())
print(test["label"].value_counts())
```

```python
# Step 2: read 5 complaints per class
train.groupby("label").head(5)
```

```python
for label in train["label"].unique():
    print("=====", label)
    for t in train[train["label"] == label]["text"].head(5):
        print("-", t[:600])
        print()
```
